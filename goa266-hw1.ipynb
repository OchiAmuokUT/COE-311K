```python
import numpy as np
import matplotlib.pyplot as plt
lotsOfDebug = False
```

```python
def monteCarloPi(sampSize):
    inCircle = 0
    
    for i in range(0,sampSize,1):
        x=np.random.rand()
        y=np.random.rand()
        distCalc = x**2 + y**2
        if (distCalc <= 1):
            inCircle += 1

    output = 4*inCircle/sampSize
    if lotsOfDebug: 
        (print('sample size of ',sampSize,' completed ---> ',output))
    return output
```

```python
def plotFunc (startVal,endVal,xStepSize):
    print('\tplotting function initalized')

    #THE LACK OF DIVISBILITY 10->1000 /20 IS HURTING ME
    if lotsOfDebug:
        print('sample size starting value is ',startVal)
        print('sample size ending value is ',endVal)
        print('step size will be ',xStepSize)
    
    xVals = []
    for i in range (startVal,endVal,20):
        xVals.append(i)

    #...so i just added a 1000 to the end
    xVals.append(endVal)  

    yVals = []
    numOfXvals = len(xVals)

    print('\t\tcalculating error values')
    
    for i in range(0,numOfXvals,1):
        yValue = np.abs(monteCarloPi(xVals[i])-np.pi)
        yVals.append(yValue)
        if lotsOfDebug:
            print('resulting in an error of ',yValue)
            
    plt.xlabel('#of Samples')
    plt.ylabel('error')
    plt.scatter(xVals,yVals,s=10)
    print('\tplotting function complete')
            
```

```python
def main():
    print('goa266-hw1 initalized!')
    #plotFunc (startVal,endVal,xStepSize):
    plotFunc(10,1000,20)
    print('goa266-hw1 complete!')
main()
```
