```python
import numpy as np
def monteCarloPi(sampSize):
    inCircle = 0
    
    for i in range(0,sampSize,1):
        x=np.random.rand()
        y=np.random.rand()
        distCalc = x**2 + y**2
        if (distCalc <= 1):
            inCircle += 1

    output = 4*inCircle/sampSize
    print(output)
    return output
```

```python
monteCarloPi(1000)
```
