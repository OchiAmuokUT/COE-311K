```python
#Leibinz Pi Estimation Challenge-2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def monteCarloPi(sampSize):
    inCircle = 0
    
    for i in range(0,sampSize,1):
        x=np.random.rand()
        y=np.random.rand()
        distCalc = x**2 + y**2
        if (distCalc <= 1):
            inCircle += 1

    output = 4*inCircle/sampSize
    return output
```

```python
def errorEstimation(calcdValue):
    trueValue = np.pi
    absError = np.abs(calcdValue-trueValue)
    relError = np.abs((calcdValue-trueValue)/trueValue)

    output = [absError,relError]
    return output
```

```python
def errorPlotting(inputArray):
    plt.xlabel('AbsoluteError')
    plt.ylabel('RelativeError')
    plt.scatter(inputArray[:,0],inputArray[:,1],s=10)
```

```python
def main():
    reps = 100
    samplSize = 1000
    arryTheArray = np.zeros((reps,2))
    for i in range (0,reps,1):
        varryTheVar = monteCarloPi(samplSize)
        arryTheArray[i] = errorEstimation(varryTheVar)
        
    errorPlotting(arryTheArray)
main()
```
